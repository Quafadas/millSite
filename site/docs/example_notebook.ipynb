# Example notebook

This page is a Jupyter notebook, `example_notebook.ipynb`, dropped into `docs/` next to the Markdown pages. It is rendered from the outputs stored in the file: nothing runs at build time. See [Notebooks](notebooks.md) for how it works.

The outputs mimic the shapes the [Almond](https://almond.sh) Scala kernel writes.


```scala
import almond.display._
```

## Values and printing

Expression results show up as plain text, with the kernel's terminal colours stripped.

```scala
case class Point(x: Double, y: Double)

val points = List(Point(0, 0), Point(3, 4), Point(6, 8))
```

```scala
points.foreach(p => println(s"distance: ${math.hypot(p.x, p.y)}"))
```

Images are written to files next to the page, rather than inlined:

```scala
Svg(barChart(commitsPerDay))
```

Vega-Lite and Plotly specs become interactive charts. Hover over a bar:

```scala
commitsPerDay.plot(_.mark.bar(tooltip = true))
```
